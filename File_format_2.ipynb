```python
CSE=pd.DataFrame(
    {
        "Name":["Ajay","Balaji"],
        "Marks":[91,95]})
ECE=pd.DataFrame(
    {
        "Name":["Anish","Karan"],
       "Marks":[56,78]})
IT=pd.DataFrame(
    {
        "Name":["Harshit","Jhansi"],
        "Marks":[76,89]})
with pd.ExcelWriter("Student_Department.xlsx") as f1:
  CSE.to_excel(f1,sheet_name="CSE",index=False)
  ECE.to_excel(f1,sheet_name="ECE",index=False)
  IT.to_excel(f1,sheet_name="IT",index=False)
```
