###Q7.1. Write a function my_abs(val) that returns the absolute value of val without using Python's built-in abs().Test it with a positive number, a negative number, and 0.

#Q7.5. Write a function swap_message(val1, val2) that swaps the two LOCAL variables inside the function andprints them. Then call swap_message(6, 3) and separately print the original variables x = 6, y = 3 after the call toshow that the originals are unaffected.

````python
def swap_message(val1, val2):
    # Swap the local variables
    temp = val1
    val1 = val2
    val2 = temp

    print("Inside the function:")
    print("val1 =", val1)
    print("val2 =", val2)


x = 6
y = 3

swap_message(x, y)

print("Outside the function:")
print("x =", x)
print("y =", y)
````

##Q8.2. Create a global variable counter = 0. Write a function show_counter() that prints counter. Call the functionand confirm it can read the global variable.

##Q8.3. Now write a function increment_counter() that tries to do counter = counter + 1 without using the globalkeyword. Run it and explain the error you get.

````python
counter = 0

def increment_counter():
    counter = counter + 1

increment_counter()
#The error occurs because Python considers counter to be a local variable when we assign a value to it inside the function. However, the statement tries to use counter before it has been assigned a local value. Therefore, Python raises an UnboundLocalError. To modify the global variable counter inside the function, we need to use the global keyword.
````

##Q8.4. Fix the function from the previous question using the global keyword so it correctly increments the globalcounter variable.

````python
counter = 0

def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print(counter)
````
