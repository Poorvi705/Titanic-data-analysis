```python
import pandas as pd

data = {
    "Name": ["Amit", "Riya", "Rahul", "Priya", "Karan",
             "Neha", "Arjun", "Sneha", "Vikram", "Anjali"],

    "Age": [20, 21, 20, 22, 21, 19, 22, 20, 21, 19],

    "Department": ["CSE", "IT", "CSE", "ECE", "IT",
                   "CSE", "ECE", "IT", "CSE", "ECE"],

    "Marks": [85, 90, 78, 88, 76, 92, 81, 87, 89, 95]
}

df = pd.DataFrame(data)
print(df)
```

```python
df.to_csv("df_csvoutput.csv",index=False)
df_csv=pd.read_csv("/content/df_csvoutput.csv")
print(df_csv.head())
```

```python
df.to_excel("df_exceloutput.xlsx",index=False)
df_excel=pd.read_excel("/content/df_exceloutput.xlsx")
print(df_excel.tail())
```

```python
df.to_json("df_jsonoutput.json",orient="records",indent=4)
df_json=pd.read_json("/content/df_jsonoutput.json")
print(df_json.head())
```

```python
df.to_parquet("df_parquetoutput.parquet",index=False)
```

```python
#Insert the dataframe
print("Shape of the dataframe:",df.shape)
print("Print 3 rows:\n",df.head(3))
print("Last 2 rows:\n",df.tail(2))
print("columns:",df.columns.tolist)
print("Data types:\n",df.dtypes)
print("Information:\n",df.info())
print("Statistics of data frame:\n",df.describe())
```

#File Formats
##1.csv
##2.excel
##3.json
##4.parquet

```python
#load the dataset in file format
Iris.to_csv("Iris_csvoutput.csv",index=False)
#Read the saved dataset
Iris_csv=pd.read_csv("/content/Iris_csvoutput.csv")
print(Iris_csv.head)
```

```python
Iris.to_excel("Iris_exceloutput.xlsx",index=False)
#load
Iris_excel=pd.read_excel("/content/Iris_exceloutput.xlsx")
print(Iris_excel.tail(10))
```

```python
#Json file format
Iris.to_json("Iris_jsonoutput.json",orient="records",indent=4)
Iris_json=pd.read_json("/content/Iris_jsonoutput.json")
Iris_json.head(5)
```

```python
#parquet file format
Iris.to_parquet("Iris_parquetoutput.parquet",index=False)
Iris_parquet=pd.read_parquet("/content/Iris_parquetoutput.parquet")
print(Iris_parquet.head(3))
```

```python
#compare the size of file format
import os
files=[
    "Iris_csvoutput.csv",
    "Iris_exceloutput.xlsx",
    "Iris_jsonoutput.json",
    "Iris_parquetoutput.parquet"
]
for file in files:
  size=os.path.getsize(file)
  print(f"{file}:{size} bytes")
```
