```python
#create a tuple
student=(101,"Sanjiv",8.9)
print(student)

#tuple using index
print("Student info in first location" ,student[0])
print(student[1])

# negative index
print("Example for negative index:", student[-2])
```

```python
subjects=("DSA","DSV","OS","PYTHON","MACHINE LEARNING")
print(len(subjects))
```

```python
#slicing
subjects=("DSV","DSA","STATISTICS","SQL","MACHINE LEARNING")
print(subjects[1:3])
print(subjects[:5])
print(subjects[4:])
```

```python
#tuple immutable
students_info=("Chirag",21,"CSE",8.3)
students_info[3]=9.1
print(students_info)
```

```python
#nested tuple
student=(101,"Rahul",(89,78,66))
print(student)
print(student[2])
print(student[2][0])
```

```python
# count function
marks=(75,78,68,75,89,75,92,98)
print("time of occurence:",marks.count(75))
```

```python
#find the element is present in tuple or not
subjects=("Python",
          "DSV",
          "SQL",
          "MACHINE LEARNING",
          "STATISTICS")
print("Python" in subjects)
print("JAVA" in subjects)
```

#SET
## unordered collection of items
## It is mutable
## allows the different data types
## it doesn't support for index
## it gives output with uniques(NO DUPLICATION IS ALLOWED IN SETS)


```python
#symbol for set: { }
#Create a set with similar items
marks={75,68,89,55,78,89,75,66,55,99}
print(marks)
```

```python
#create a empty set
a={}
a=set()
#check its type
print(type(a))
```

```python
# set doesn't supporting index
students={"Anit","Riya","Priya"}
print(students[1])
```

```python
# add an item in the set
students={"Amit","Kiran","Sneha"}
students.add("Priya")
print(students)
students.update(["Ram","Chirag","Vani","Devi"])
print(students)
```

```python
# remove an item from the set
students={"Amit","Riya","Priya"}
students.remove("Priya")
print(students)

students.discard("Amit")
print(students)

```

```python
students={"Amit","Riya","Priya"}
students.remove("Karan")
print(students)
```

```python
students={"Amit","Riya","Priya"}
students.discard("Karan")
print(students)
```

```python
# create a set and clear list of items
students_info={"Amit",
               "Karan",
               "Riya"}
print(students_info)
students_info.clear()
print(students_info)
```

```python
marks={99,68,77,66}
del marks
print(marks)
```

```python
#set operations
python_students={"Amit","Neha","Sanjiv","Manish"}
java_students={"Amit","Karan","Dilip","Manish","Neha"}

#for union operation by using set
result=python_students.union(java_students)
print("Union operation:",result)

#for intersection operation by using set
result_intersection=python_students.intersection(java_students)
print("Intersection Operation:", result_intersection)

#to find difference with two set
res_diff=python_students.difference(java_students)
print("Difference Operation:",res_diff)
```

```python
# convert list to set
stud_id=[101,102,103,102,103,104,105]
unique_stud_id=set(stud_id)
print(unique_stud_id)
uniq_roll_no=list(unique_stud_id)
print(uniq_roll_no)
```
